## Question 3

```python
import random
import math

random.seed(1024170120)
numbers = [random.randint(100, 900) for _ in range(100)]
print(numbers)
```

```python
odd_count = sum(1 for number in numbers if number % 2 != 0)
print("Number of odd values:", odd_count)
```

```python
even_count = sum(1 for number in numbers if number % 2 == 0)
print("Number of even values:", even_count)
```

```python
def is_prime(number):
    if number < 2:
        return False
    for divisor in range(2, math.isqrt(number) + 1):
        if number % divisor == 0:
            return False
    return True
```

```python
prime_values = [number for number in numbers if is_prime(number)]
print("Number of prime values:", len(prime_values))
print("Prime values:", prime_values)
```

```python
frequency = {}
for number in numbers:
    frequency[number] = frequency.get(number, 0) + 1

most_common = max(frequency, key=frequency.get)
print("Most frequent value:", most_common)
print("Frequency:", frequency[most_common])
```
