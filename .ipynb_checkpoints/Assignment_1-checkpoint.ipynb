#### 5.3

```python
def check_prime(value):
    if value < 2:
        return False

    for divisor in range(2, int(value ** 0.5) + 1):
        if value % divisor == 0:
            return False

    return True

limit = int(input("Enter number: "))
prime_total = 0

for value in range(2, limit + 1):
    if check_prime(value):
        prime_total += value

print("Sum of prime numbers up to", limit, "is", prime_total)
```

#### 6.1

```python
def sum_odd_numbers(limit):
    total = 0

    for value in range(1, limit + 1, 2):
        total += value

    return total

limit = int(input("Enter the number: "))
print("Sum of odd numbers up to", limit, "is", sum_odd_numbers(limit))
```

#### 6.2

```python
def check_prime(value):
    if value < 2:
        return False

    for divisor in range(2, int(value ** 0.5) + 1):
        if value % divisor == 0:
            return False

    return True

limit = int(input("Enter number: "))
prime_total = sum(
    value for value in range(2, limit + 1)
    if check_prime(value)
)

print("Sum of prime numbers up to", limit, "is", prime_total)
```
