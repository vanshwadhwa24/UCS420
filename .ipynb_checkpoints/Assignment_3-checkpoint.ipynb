```python
#6
data_emp = {
    "Employee_ID": [101, 102, 103, 104, 105],
    "Name": ["Alice", "Bob", "Charlie", "Diana", "Edward"],
    "Department": ["HR", "IT", "IT", "Marketing", "Sales"],
    "Age": [29, 34, 41, 28, 38],
    "Salary": [50000, 70000, 65000, 55000, 60000],
    "Years_of_Experience": [4, 8, 10, 3, 12],
    "Joining_Date": ["2020-03-15", "2017-07-19", "2013-06-01", "2021-02-10", "2010-11-25"],
    "Gender": ["Female", "Male", "Male", "Female", "Male"],
    "Bonus": [5000, 7000, 6000, 4500, 5000],
    "Rating": [4.5, 4.0, 3.8, 4.7, 3.5]
}

emp = pd.DataFrame(data_emp)
print(emp)
emp.to_csv("employee.csv", index=False)
```

```python
#6.f
emp = emp.sort_values(by="Salary", ascending=False)
print(emp)
```

```python
#6.g
def pf_cat(rating):
    if rating >= 4.5:
        return "excellent"
    elif rating >= 4.0:
        return "Good"
    else:
        return "Average"

emp["Performance"] = emp["Rating"].apply(pf_cat)
```

```python
#6.h
print(emp.isnull().sum())
```

```python
#6.i
emp=emp.rename(columns={"Employee_ID":"ID"})
print(emp.columns)
```

```python
#6.j.i
experienced=emp[emp["Years_of_Experience"]>5]
print(experienced)
```

```python
#6.j.ii
it_emp=emp[emp["Department"]=="IT"]
print(it_emp)
```

```python
#6.k
emp["Tax"]=emp["Salary"]*0.10
print(emp["Tax"])
```

```python
#6.l
emp.to_csv("updated-employee.csv", index=False)
```
