```python
steps = np.array([ [5000, 6200, 7100], [8000, 7500, 9000], [4500, 5100, 4800], [9000, 8500, 9500]]) 
print("total steps",np.sum(steps))
print("Mean steps", np.mean(steps))
print("Maximum", np.max(steps))
print("Minimum", np.min(steps))
print("Total for each day", np.sum(steps, axis=0))
print("Total for each user", np.sum(steps, axis=1))
position = np.unravel_index(np.argmax(steps), steps.shape)
print("Position of maximum:", position)
print("Maximum steps:", steps[position])
```

```python
X = np.array([ [6, 70, 3], [5, 50, 6], [8, 80, 2], [4, 30, 8] ]) 
y = np.array([40, 65, 30, 85])

print("Shape", X.shape)
print("Dimensions", X.ndim)

print("transpose",X.T)
print("Transpose swaps rows and columns.After transpose, X.T is 3 × 4,each row of X.T contains all values of one feature across the 4 users. ")

XtX = X.T @ X
print("matrix product",XtX)

inverse = np.linalg.inv(XtX)
print("matrix inverse",inverse)

beta = np.linalg.inv(X.T @ X) @ X.T @ y
print("equation is",beta)

print("\nCoefficients:")
print("Sleep:", beta[0])
print("Activity:", beta[1])
print("Stress:", beta[2])

new_user = np.array([5, 40, 7])
prediction = new_user @ beta
print("Predicted assistance score:", prediction)
```

question4 f:
The coefficients show the effect of each feature on the assistance score while keeping the other features constant.

Sleep: -0.159 → slight decrease in score
Activity: +0.141 → slight increase in score
Stress: +10.063 → increase in score
